# Importing Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

# Defingin funtion

```python
def compute_convolution(x, h):
    N = len(x)
    M = len(h)
    z = np.zeros(N + M - 1)  # full convolution length
    
    for n in range(N + M - 1):
        s = 0
        for k in range(N):
            h_index = n - k
            if 0 <= h_index < M:
                s += x[k] * h[h_index]
        z[n] = s
    
    # Normalize
    max_val = np.max(np.abs(z))
    if max_val != 0:
        z = z / max_val
    return z
```

# Creating rec signal

```python
sr = 100
t = np.linspace(0, 5, 5*sr, endpoint=False)
x = np.where(np.abs(t-2.5) <= 0.5, 1.0, 0.0)
```

# Plotting the convolved signal, thrice

```python
z1 = compute_convolution(x, x)
z2 = compute_convolution(z1, x)
z3 = compute_convolution(z2, x)
# Time vectors for plotting
t1 = np.linspace(0, 5, len(z1), endpoint=False)
t2 = np.linspace(0, 5, len(z2), endpoint=False)
t3 = np.linspace(0, 5, len(z3), endpoint=False)

plt.figure(figsize=(12,6))
plt.plot(t, x, label='x[n]')
plt.plot(t1, z1, label='x*x')
plt.plot(t2, z2, label='x*x*x')
plt.plot(t3, z3, label='x*x*x')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')
plt.title('Repeated Convolution of x[n]')
plt.legend()
plt.grid(True)
plt.show()
```

Z smoothens aftern each convultion it is a rect function originally. Turns into a trapezoid after first convolution. It is smoother like a quadratic function on second convolution. It is much smoother on the third convolution.
